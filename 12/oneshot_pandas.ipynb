```python
df3[df3['Age']>30 & df3['City']=='Mumbai']
```

Your logic is fine. Your syntax is wrong. Pandas isn’t Python here—it’s NumPy boolean indexing, and you violated its rules.

The error
df3[df3['Age']>30 & df3['City']=='Mumbai']

What’s wrong (exactly)

Operator precedence
& runs before > and ==.
So Python reads it like this (which is garbage):

df3['Age'] > (30 & df3['City']) == 'Mumbai'


That makes no sense → ValueError / TypeError

Missing parentheses
Every condition in pandas boolean indexing MUST be wrapped in parentheses.

✅ Correct code
df3[(df3['Age'] > 30) & (df3['City'] == 'Mumbai')]

Rule you must remember (no excuses)

Use & not and

Use | not or

ALWAYS wrap each condition in parentheses

❌ Wrong

df3['Age'] > 30 & df3['City'] == 'Mumbai'


✅ Right

(df3['Age'] > 30) & (df3['City'] == 'Mumbai')

Bonus: common variations
# OR condition
df3[(df3['Age'] > 30) | (df3['City'] == 'Mumbai')]

# NOT condition
df3[~(df3['City'] == 'Mumbai')]

```python
employees = pd.DataFrame({
'employee_id': [1, 2, 3, 4, 5],
'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})
# DataFrame 2: Salary information
salaries = pd.DataFrame({
'employee_id': [1, 2, 3, 6, 7],
'salary': [60000, 80000, 65000, 70000, 90000],
'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)
```

```python
pd.merge(employees, salaries, on='employee_id', how='outer')
# Meaning of on --> The on parameter specifies the common column(s) to join on. In this case, both DataFrames have an employee_id column, so we use that as the key for merging.
# Meaning of how --> The how parameter determines the type of merge to be performed. An outer join includes all records from both DataFrames, filling in NaN for missing matches.
```

```python
pd.merge(employees, salaries, on='employee_id', how='left')
# left means all records from the left DataFrame (employees) and matching records from the right DataFrame (salaries). If there is no match, NaN values are filled in for the right DataFrame's columns.
```

```python
pd.merge(employees, salaries, on='employee_id', how='right')
# right means all records from the right DataFrame (salaries) and matching records from the left DataFrame (employees). If there is no match, NaN values are filled in for the left DataFrame's columns.
```
